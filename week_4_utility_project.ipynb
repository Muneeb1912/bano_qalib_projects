```python
%%writefile utility.py
import math

def factorial(number):
    result = 1
    
    for val in range(1,number+1):
        result*= val
    
    return result

def prime_check(number):
    if number<2:
        return False
    for value in range(2,number):
        if number%value == 0:
            return False
    return True

def square_root(number):
    return math.sqrt(number)
```
