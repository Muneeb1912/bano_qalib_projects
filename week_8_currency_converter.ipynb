# currency_converter_project

```python
import requests
print("===== Currency Converter =====")

try:
    amount = float(input("Enter Currency: "))
    from_currency = input("Convert from (example USD: )").strip().upper()
    to_currency = input("Convert to (example PKR: )").strip().upper()

    if amount < 0:
        print("Amount cannot be negative.")
    elif from_currency == "" or to_currency == "":
        print("currency code cannot be empty.")
    elif from_currency == to_currency:
        print(f"Converted Amount: {amount} {to_currency}")
    else:
        url = f"https://api.frankfurter.dev/v2/rate/{from_currency}/{to_currency}"
        respone = requests.get(url, timeout=10)
        if respone.status_code == 200:
            data = respone.json()
            exchange_rate = data["rate"]
            converted_amount = amount*exchange_rate

            print("\n===== Conversion Result =====")
            print(f"Rate date: {data["date"]}")
            print(f"1 {from_currency} = {exchange_rate} {to_currency}")
            print(f"{amount} {from_currency} = {round(converted_amount,2)} {to_currency}")
        else:
            print("Currency Code was not found.")
    
except ValueError:
    print("Please enter a valid numeric amount.")
except request.exceptions.RequestException:
    print("Internet or API connection error.")
```
