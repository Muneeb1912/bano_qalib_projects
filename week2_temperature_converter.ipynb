## Temperature Converter
Convert Celsius to Fahrenheit and Fahrenheit to Celsius.

### Formulas
Fahrenheit = (Celsius * 9 / 5) + 32
Celsius = (Fahrenheit - 32) * 5 / 9

```python
print("Temperature Convertor")
# celcius to fahrenheit
cel_given = float(input("Enter temp in Celcius : "))
fah_ans = (cel_given * 9/5)+32
print(f"{cel_given} celcius is equal to {fah_ans} fahrenheit")

# fahrenheit to celcius
fah_given = float(input("Enter temp in Celcius : "))
cel_ans = (fah_given-32)*5/9
print(f"{fah_given} celcius is equal to {cel_ans} fahrenheit")
```
