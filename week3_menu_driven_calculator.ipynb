```python
print("===== Menu Driven Calculator =====")
while True:
    # menu choices
    print("\n Choose an options")
    print("+ = addition")
    print("- = subtraction")
    print("* = multiplication")
    print("/ = division")
    print("0 = Exit")

    sel_opt = input("Enter your choice : ")
    
    if sel_opt == "0":
        print("Calculator Closed")
        break
    if sel_opt != "+" and sel_opt != "-" and sel_opt != "*" and sel_opt != "/":
        print("Invalid Choice")
        continue
    else:
        num1 = float(input("Enter First number :"))
        num2 = float(input("Enter Second number :"))

        if sel_opt == "+":
            ans = num1 + num2
            print(f"Addition of {num1} and {num2} is : {ans}")
        
        elif sel_opt == "-":
            ans = num1 - num2
            print(f"Subtraction of {num1} and {num2} is : {ans}")
        
        elif sel_opt == "*":
            ans = num1 * num2
            print(f"Multiplication of {num1} and {num2} is : {ans}")
        
        elif sel_opt == "/":
            if num2 == 0:
                print("Math Error not divide when 0 is in demoninator ")
                continue
            else :
                ans = num1 / num2
                print(f"Division of {num1} and {num2} is : {ans}")
```
