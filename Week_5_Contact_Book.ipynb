# Week_5_Contact_Book

```python
### contacts = {}
print("===== Contact Book =====\n")
while True:
    print("\n1. Add Contact")
    print("\n2. Search Contact")
    print("\n3. Exit")

    choice = input("Enter Your Choice: ").strip()
    if choice == "1":
        name = input("Enter Contact Name: ").strip().title()
        phone = input("Enter Phone Number: ").strip()
        email = input("Enter Your Email: ").strip()
        if name == "" or phone == "" or email == "":
            print("All Information Is required.")
        else:
            contacts[name]= {
                "phone": phone,
                "email": email
            }
        print("Contact Add Successfully.")
    elif choice == "2":
        search_name = input("Enter Contact Name to Search: ").strip().title()
        if search_name in contacts:
            print("\n Contact Found")
            print(f"Name: {search_name}")
            print(f"Phone: {contacts[search_name]["phone"]}")
            print(f"Email: {contacts[search_name]["email"]}")
        else:
            print("Contact not Found.")

    elif choice == "3":
        print("Contact Book Closed.")
        break
    else:
        print("Invalid choice enter 1, 2, or 3.")   
```
