```python
# Week 1 Project: Simple Calculator
# This notebook takes two numbers and performs addition, subtraction, multiplication, and
# division.
```

```python
print("===== Simple Calculator ===== \n")
#take two inputs from user
num_1 = float(input("Enter First Number : "))
num_2 = float(input("Enter Second Number (Not Zero): "))

# perform calculations 
addition = num_1 + num_2
subtraction = num_1 - num_2
multiplication = num_1 * num_2
division = num_1 / num_2

# print results
print("\n===== Results ===== \n")
print(f"Addition of {num_1} and num_2 is : {addition}")
print(f"Subtraction of {num_1} and num_2 is : {subtraction}")
print(f"Multiplication of {num_1} and num_2 is : {multiplication}")
print(f"Division of {num_1} and num_2 is : {division}")
```
